```python
import xarray as xr
from pathlib import Path
from ecod_calculation import cod_from_water_path
from utils_cartopy import plot_global_field, plot_north_pole_field
```

```python
output_dir = Path('ecod_validation_slides')
output_dir.mkdir(exist_ok=True)

path_era5 = Path('data/era5')
```

```python
era5_paths = path_era5.glob("era5_single_levels_monthly_*.nc")
#ecod_paths = path_era5.glob("era5_true_tcc_ecod_monthly*.nc")
```

```python
ds_era5 = xr.open_mfdataset(era5_paths)
#ds_ecod = xr.open_mfdataset(ecod_paths)
```

```python
cod = cod_from_water_path(ds_era5.tclw * 1000, ds_era5.tciw * 1000)
lon = cod.longitude
lat = cod.latitude
```

```python
date = '2015-09'

plot_global_field(
    ds_era5.tciw.sel(date=date)[0] * 1000, 
    lon, 
    lat, 
    f'tciw, {date}', 
    output_dir / 'tciw.png', vmin=0, vmax=120, label = 'kg/m^2',
    cmap='Blues')

plot_global_field(
    ds_era5.tclw.sel(date=date)[0] * 1000, 
    lon, 
    lat, 
    f'tclw, {date}', 
    output_dir / 'tclw.png', vmin=0, vmax=200, label = 'kg/m^2',
    cmap='Blues')

plot_global_field(
    cod.sel(date=date)[0], 
    lon, 
    lat, 
    f'cod, {date}', 
    output_dir / 'cod.png', vmin=0, vmax=50, label = '',
    cmap='Blues')

plot_global_field(
    (cod * ds_era5.tcc).sel(date=date)[0], 
    lon, 
    lat, 
    f'ecod, {date}', 
    output_dir / 'ecod.png', vmin=0, vmax=50, label = '',
    cmap='Blues')

plot_global_field(
    cod.sel(date=date)[0], 
    lon, 
    lat, 
    f'cod, {date}', 
    output_dir / 'cod5.png', vmin=0, vmax=10, label = '',
    cmap='Blues')

plot_global_field(
    (cod * ds_era5.tcc).sel(date=date)[0], 
    lon, 
    lat, 
    f'ecod, {date}', 
    output_dir / 'ecod5.png', vmin=0, vmax=10, label = '',
    cmap='Blues')
```

```python
plot_north_pole_field(
    ds_era5.tciw.sel(date=date)[0] * 1000, 
    lon, 
    lat, 
    f'tciw, {date}', 
    output_dir / 'np_tciw.png', vmin=0, vmax=120, label = 'kg/m^2',
    cmap='Blues')

plot_north_pole_field(
    ds_era5.tclw.sel(date=date)[0] * 1000, 
    lon, 
    lat, 
    f'tclw, {date}', 
    output_dir / 'np_tclw.png', vmin=0, vmax=200, label = 'kg/m^2',
    cmap='Blues')

plot_north_pole_field(
    cod.sel(date=date)[0], 
    lon, 
    lat, 
    f'cod, {date}', 
    output_dir / 'np_cod.png', vmin=0, vmax=50, label = '',
    cmap='Blues')

plot_north_pole_field(
    (cod * ds_era5.tcc).sel(date=date)[0], 
    lon, 
    lat, 
    f'ecod, {date}', 
    output_dir / 'np_ecod.png', vmin=0, vmax=50, label = '',
    cmap='Blues')

plot_north_pole_field(
    cod.sel(date=date)[0], 
    lon, 
    lat, 
    f'cod, {date}', 
    output_dir / 'np_cod5.png', vmin=0, vmax=10, label = '',
    cmap='Blues')

plot_north_pole_field(
    (cod * ds_era5.tcc).sel(date=date)[0], 
    lon, 
    lat, 
    f'ecod, {date}', 
    output_dir / 'np_ecod5.png', vmin=0, vmax=10, label = '',
    cmap='Blues')
```
